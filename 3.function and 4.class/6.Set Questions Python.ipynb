```python
def Discount():
    price=int(input("Enter the amount:"))
    if price>=100:
      
        disc=20/100
        caldiscount=(price*disc)+price
        return "After Discount:$=",caldiscount
        
    else:
        return "Not qualified"
discount_price=Discount()
print(discount_price)
```

```python
def leapyear():
    year=int(input("Enter the amount:"))
    if (year%4)==0 and (year%100)!=0 or (year%400)==0:
        return "Year is Leap year " 
    
    else:
        return "Year is not Leap year"
leapyear=leapyear()
print(leapyear)
```

```python
def prime():
    num = int(input("Enter Number:"))
    count = 0
    
    for i in range(1, num + 1):
        if num % i == 0:
            count = count + 1
    
    if count == 2:
        return i , "is Prime Number"
    else:
        return "Not a Prime Number"
primenum=prime()
print(primenum)
```
