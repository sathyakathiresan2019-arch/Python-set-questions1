```python
def Primenumber():
    num = int(input("Enter a number: "))
    if num < 2:
        print("Not Prime")
    else:
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                return "Not Prime"
                break
            else:
                return "Prime"
prime=Primenumber()
print(prime)
```

```python
def factorial(num):
    if num == 0 or num == 1:
        return 1
    else:
        return num * factorial(num - 1)
num = int(input("Enter a number: "))
print(factorial(num))
```

```python
def Anagram():
    string1 = input("Enter first string: ")
    string2 = input("Enter second string: ")
    string1 = string1.lower()
    string2 = string2.lower()
    if sorted(string1) == sorted(string2):
        return "The strings are anagrams"
    else:
        return "The strings are not anagrams"
Anagramwords=Anagram()
print(Anagramwords)
```

```python
def fibonacci(n):
    a = 0
    b = 1
    series = []

    for i in range(n):
        series.append(a)

        c = a + b
        a = b
        b = c

    return series


result = fibonacci(10)
print(result)
```

```python
def palindrome(word):
    
    reverse = word[::-1]

    if word == reverse:
        return word + " is a palindrome"
    else:
        return word + " is not a palindrome"


result = palindrome("madam")
print(result)
result = palindrome("level")
print(result)
result = palindrome("malayalam")
print(result)
result = palindrome("hello")
print(result)
```
