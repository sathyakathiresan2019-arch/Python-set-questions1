```python
def Passcode():
    num = int(input("Enter a number: "))
    digit1=num%10 
    num2=num//10  
    digit2=num2%10 
    num3=num2//10  
    digit3=num3%10 
    digit4=num3//10
    sum=digit1+digit2+digit3+digit4
    return "Sum of digits:", sum
DigiPasscode=Passcode()
print(DigiPasscode)
```

```python
def PasscodeNumber():
    num = int(input("Enter a number: "))
    
    sum_digits = 0
    
    while num > 0:
        digit = num % 10     
        sum_digits = sum_digits + digit
        num = num // 10     
    
    return "Sum of digits:", sum_digits
DigiPasscodes=PasscodeNumber()
print(DigiPasscodes)
```

```python
def prime():
    num = int(input("Enter Number:"))
    count = 0
    for i in range(1, num + 1):
        if num % i == 0:
            count = count + 1
     if count == 2:
        return i , "is Prime Number"
    else:
        return "Not a Prime Number"
primenum=prime()
print(primenum)
```

```python
def Primenumber():
    num = int(input("Enter a number: "))
    if num < 2:
        print("Not Prime")
    else:
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                return "Not Prime"
                break
            else:
                return "Prime"
prime=Primenumber()
print(prime)
```

```python
def factorial(num):
    if num == 0 or num == 1:
        return 1
    else:
        return num * factorial(num - 1)
num = int(input("Enter a number: "))
print(factorial(num))
```

```python
def armstrong():
    
    num=int(input("Enter a number:"))
    temp=num
    sum=0
    digits=len(str(num))
    while temp>0:
        digit=temp%10    
        sum=sum+digit**digits 
        temp=temp//10   
    if sum==num:
        return"Armstrong Number"
    else:
        return"Not Armstrong Number"      
arm=armstrong()
print(arm)
```

```python
def swapped():
    text=input("Enter a text:")
    if len(text)<=1:
        return text
    else:
        swapped= text[-1]+text[1:-1]+text[0]
        return swapped
swappingnum=swapped()
print(swappingnum)
```
